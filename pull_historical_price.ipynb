```python
# ---- fetch ----

client1 = FMPClient()

price_frames = []
failed_tickers = {}                               # ticker -> reason

for symbol in all_tickers:
    try:
        raw = client1.get_data('historical-price-eod', symbol)
        if not raw or isinstance(raw, dict):
            raise ValueError(f"bad response: {raw}")
        df = pd.DataFrame(raw)
        df['symbol'] = symbol
        price_frames.append(df)
    except Exception as e:
        failed_tickers[symbol] = f"{type(e).__name__}: {e}"

print(f"OK: {len(price_frames)}  |  failed: {len(failed_tickers)}")

all_tickers_price = pd.concat(price_frames, ignore_index=True)
all_tickers_price['date'] = pd.to_datetime(all_tickers_price['date'])
all_tickers_price.to_csv('all_tickers_price.csv', index=False)

# ---- results ----

overlap_df = pd.DataFrame([
    {'symbol': t, 'years': ','.join(map(str, y)), 'n_years': len(y)}
    for t, y in overlapped.items()
]).sort_values(['n_years', 'symbol'], ascending=[False, True])

failed_df = pd.DataFrame([
    {'symbol': s, 'reason': r, 'years': ','.join(map(str, sorted(membership[s])))}
    for s, r in failed_tickers.items()
])

print(overlap_df.head(20).to_string(index=False))
print(failed_df.to_string(index=False))
```
